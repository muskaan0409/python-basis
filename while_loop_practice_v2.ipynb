Q1. Print numbers from 1 to 20 using while loop.

- Initialize a variable

- Increment inside loop

Q4. Find sum of numbers from 1 to n.

Approach:
- Take input n
- Add numbers in loop

```python
#write answer here
m=0
sum=0
n=int(input("enter the number from where youy want to add:"))
while m!=(n+1):
  sum=sum+m
  m=m+1

print(f"total sum is :{sum}")
```

Q5. Reverse a number using while loop.

Approach:
- Extract last digit
- Build reversed number

```python
#write answer here

var= int(input())

new = var

res=0

while var>0:
  temp = var % 10
  res = res*10 + temp
  var = var // 10

print(res)


if new==res:
  print("Palindrome")
else:
  print("Not Palindrome")


```
